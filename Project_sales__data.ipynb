**# Introduction**

The **Sales Data Analysis Project** aims to analyze sales transaction data from 2022 to 2025 to understand business performance, sales trends, profitability, and customer purchasing behavior. The dataset contains information about customers, products, sales transactions, order dates, quantities sold, discounts, costs, profits, payment methods, delivery details, and product returns.

In this project, Python is used to analyze and interpret the dataset with the help of three important libraries: **Pandas, NumPy, and Matplotlib**. Pandas is used for data exploration, cleaning, and grouping. NumPy is used to perform numerical calculations and statistical analysis. Matplotlib is used to create graphs and visualizations that make the results easier to understand.

The analysis focuses on key business metrics such as total sales, total profit, profit margin, average order value, and the number of customers and orders. It also examines yearly sales performance, product category performance, top-selling products, sales channels, customer segments, country-wise sales, product returns, monthly sales trends, discount effectiveness, and delivery performance.

The main objective of this project is to transform raw sales data into meaningful business insights that can help identify strengths, recognize areas for improvement, and support better data-driven business decisions.

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("/content/drive/MyDrive/Data Analyst/Data Sets/Sales_transactions_2022_2025.csv")

print(df)
````

````python



# 3. UNDERSTAND THE DATA USING PANDAS

print("First 5 rows:")
print(df.head())

print("\nDataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
print(df.describe())

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
````

````python
# 4. DATA CLEANING USING PANDAS

# Create a copy so that the original data remains unchanged.
df = df.drop_duplicates().copy()

# Convert Order_Date to datetime format.
df["Order_Date"] = pd.to_datetime(
    df["Order_Date"], errors="coerce"
)

# Convert important numerical columns to numeric format.
numeric_columns = [
    "Customer_Age",
    "Quantity",
    "Unit_Price",
    "Discount_Percentage",
    "Sales_Amount",
    "Cost_Amount",
    "Profit",
    "Delivery_Days",
    "Customer_Rating",
    "Inventory_Level"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Remove unwanted spaces and standardize category labels.
category_columns = [
    "Customer_Gender",
    "Customer_Segment",
    "Sales_Channel",
    "Country",
    "Product_Category",
    "Product_Subcategory",
    "Payment_Method",
    "Order_Status",
    "Shipping_Method",
    "Return_Flag"
]

for col in category_columns:
    df[col] = df[col].astype("string").str.strip()

# Standardize capitalization so that Electronics and electronics
# are treated as the same category.
for col in [
    "Customer_Gender",
    "Customer_Segment",
    "Sales_Channel",
    "Product_Category",
    "Product_Subcategory",
    "Payment_Method",
    "Order_Status",
    "Shipping_Method",
    "Return_Flag"
]:
    df[col] = df[col].str.title()

print("\nDataset after cleaning:")
print(df.head())

print("\nMissing values after cleaning:")
print(df.isnull().sum())

````

````python
# 5. NUMERICAL ANALYSIS USING NUMPY
# ============================================================

# Convert sales, profit, and quantity columns to NumPy arrays.
sales = df["Sales_Amount"].to_numpy(dtype=float)
profit = df["Profit"].to_numpy(dtype=float)
quantity = df["Quantity"].to_numpy(dtype=float)

# Ignore missing values in numerical calculations.
sales_valid = sales[~np.isnan(sales)]
profit_valid = profit[~np.isnan(profit)]
quantity_valid = quantity[~np.isnan(quantity)]

print("\nNUMPY ANALYSIS")
print("-" * 50)

print("Average sales:", np.mean(sales_valid))
print("Median sales:", np.median(sales_valid))
print("Maximum sales:", np.max(sales_valid))
print("Minimum sales:", np.min(sales_valid))
print("Sales standard deviation:", np.std(sales_valid))

print("Average profit:", np.mean(profit_valid))
print("Total quantity sold:", np.sum(quantity_valid))

# Calculate overall profit margin.
total_sales = np.sum(sales_valid)
total_profit = np.sum(profit_valid)

profit_margin = np.divide(
    total_profit,
    total_sales,
    out=np.array(np.nan),
    where=total_sales != 0
) * 100

print("Overall profit margin:", round(float(profit_margin), 2), "%")

# Create a high-sales flag using the median sales value.
median_sales = np.median(sales_valid)

df["High_Sales"] = np.where(
    df["Sales_Amount"] >= median_sales,
    "High",
    "Low"
)

print("\nHigh-sales and low-sales transaction counts:")
print(df["High_Sales"].value_counts())

````

````python
# 6. KEY PERFORMANCE INDICATORS (KPIs)
# ============================================================

total_sales = df["Sales_Amount"].sum()
total_cost = df["Cost_Amount"].sum()
total_profit = df["Profit"].sum()
total_units = df["Quantity"].sum()

total_transactions = df["Transaction_ID"].nunique()
total_orders = df["Order_ID"].nunique()
total_customers = df["Customer_ID"].nunique()

profit_margin = (
    total_profit / total_sales * 100
    if total_sales != 0 else np.nan
)

average_order_value = (
    df.groupby("Order_ID")["Sales_Amount"].sum().mean()
)

print("\nKEY PERFORMANCE INDICATORS")
print("-" * 50)

print("Total Sales:", round(total_sales, 2))
print("Total Cost:", round(total_cost, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Units Sold:", total_units)
print("Unique Transactions:", total_transactions)
print("Unique Orders:", total_orders)
print("Unique Customers:", total_customers)
print("Profit Margin (%):", round(profit_margin, 2))
print("Average Order Value:", round(average_order_value, 2))
````

````python
df["Year"] = df["Order_Date"].dt.year

yearly_sales = df.groupby("Year")["Sales_Amount"].sum()
yearly_profit = df.groupby("Year")["Profit"].sum()

yearly_summary = pd.DataFrame({
    "Sales": yearly_sales,
    "Profit": yearly_profit
})

yearly_summary["Profit_Margin_%"] = (
    yearly_summary["Profit"] /
    yearly_summary["Sales"] * 100
)

yearly_summary["Sales_Growth_%"] = (
    yearly_summary["Sales"].pct_change() * 100
)

print("\nYEARLY SALES AND PROFIT")
print(yearly_summary.round(2))

best_year = yearly_sales.idxmax()
print("Insight: Highest-sales year:", best_year)
````

````python

yearly_summary[["Sales", "Profit"]].plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Yearly Sales and Profit (2022–2025)")
plt.xlabel("Year")
plt.ylabel("Amount")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
````

````python


# 4. DATA CLEANING USING PANDAS

# Create a copy so that the original data remains unchanged.
df = df.drop_duplicates().copy()

# Convert Order_Date to datetime format.
df["Order_Date"] = pd.to_datetime(
    df["Order_Date"], errors="coerce"
)

# Convert important numerical columns to numeric format.
numeric_columns = [
    "Customer_Age",
    "Quantity",
    "Unit_Price",
    "Discount_Percentage",
    "Sales_Amount",
    "Cost_Amount",
    "Profit",
    "Delivery_Days",
    "Customer_Rating",
    "Inventory_Level"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Remove unwanted spaces and standardize category labels.
category_columns = [
    "Customer_Gender",
    "Customer_Segment",
    "Sales_Channel",
    "Country",
    "Product_Category",
    "Product_Subcategory",
    "Payment_Method",
    "Order_Status",
    "Shipping_Method",
    "Return_Flag"
]

for col in category_columns:
    df[col] = df[col].astype("string").str.strip()

# Standardize capitalization so that Electronics and electronics
# are treated as the same category.
for col in [
    "Customer_Gender",
    "Customer_Segment",
    "Sales_Channel",
    "Product_Category",
    "Product_Subcategory",
    "Payment_Method",
    "Order_Status",
    "Shipping_Method",
    "Return_Flag"
]:
    df[col] = df[col].str.title()

print("\nDataset after cleaning:")
print(df.head())

print("\nMissing values after cleaning:")
print(df.isnull().sum())


# ============================================================
# 5. NUMERICAL ANALYSIS USING NUMPY
# ============================================================

# Convert sales, profit, and quantity columns to NumPy arrays.
sales = df["Sales_Amount"].to_numpy(dtype=float)
profit = df["Profit"].to_numpy(dtype=float)
quantity = df["Quantity"].to_numpy(dtype=float)

# Ignore missing values in numerical calculations.
sales_valid = sales[~np.isnan(sales)]
profit_valid = profit[~np.isnan(profit)]
quantity_valid = quantity[~np.isnan(quantity)]

print("\nNUMPY ANALYSIS")
print("-" * 50)

print("Average sales:", np.mean(sales_valid))
print("Median sales:", np.median(sales_valid))
print("Maximum sales:", np.max(sales_valid))
print("Minimum sales:", np.min(sales_valid))
print("Sales standard deviation:", np.std(sales_valid))

print("Average profit:", np.mean(profit_valid))
print("Total quantity sold:", np.sum(quantity_valid))

# Calculate overall profit margin.
total_sales = np.sum(sales_valid)
total_profit = np.sum(profit_valid)

profit_margin = np.divide(
    total_profit,
    total_sales,
    out=np.array(np.nan),
    where=total_sales != 0
) * 100

print("Overall profit margin:", round(float(profit_margin), 2), "%")

# Create a high-sales flag using the median sales value.
median_sales = np.median(sales_valid)

df["High_Sales"] = np.where(
    df["Sales_Amount"] >= median_sales,
    "High",
    "Low"
)

print("\nHigh-sales and low-sales transaction counts:")
print(df["High_Sales"].value_counts())


# ============================================================
# 6. KEY PERFORMANCE INDICATORS (KPIs)
# ============================================================

total_sales = df["Sales_Amount"].sum()
total_cost = df["Cost_Amount"].sum()
total_profit = df["Profit"].sum()
total_units = df["Quantity"].sum()

total_transactions = df["Transaction_ID"].nunique()
total_orders = df["Order_ID"].nunique()
total_customers = df["Customer_ID"].nunique()

profit_margin = (
    total_profit / total_sales * 100
    if total_sales != 0 else np.nan
)

average_order_value = (
    df.groupby("Order_ID")["Sales_Amount"].sum().mean()
)

print("\nKEY PERFORMANCE INDICATORS")
print("-" * 50)

print("Total Sales:", round(total_sales, 2))
print("Total Cost:", round(total_cost, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Units Sold:", total_units)
print("Unique Transactions:", total_transactions)
print("Unique Orders:", total_orders)
print("Unique Customers:", total_customers)
print("Profit Margin (%):", round(profit_margin, 2))
print("Average Order Value:", round(average_order_value, 2))


# ============================================================
# 7. INSIGHT 1: YEARLY SALES AND PROFIT
# ============================================================

df["Year"] = df["Order_Date"].dt.year

yearly_sales = df.groupby("Year")["Sales_Amount"].sum()
yearly_profit = df.groupby("Year")["Profit"].sum()

yearly_summary = pd.DataFrame({
    "Sales": yearly_sales,
    "Profit": yearly_profit
})

yearly_summary["Profit_Margin_%"] = (
    yearly_summary["Profit"] /
    yearly_summary["Sales"] * 100
)

yearly_summary["Sales_Growth_%"] = (
    yearly_summary["Sales"].pct_change() * 100
)

print("\nYEARLY SALES AND PROFIT")
print(yearly_summary.round(2))

best_year = yearly_sales.idxmax()
print("Insight: Highest-sales year:", best_year)


# GRAPH 1: YEARLY SALES AND PROFIT

yearly_summary[["Sales", "Profit"]].plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Yearly Sales and Profit (2022–2025)")
plt.xlabel("Year")
plt.ylabel("Amount")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# ============================================================
# 8. INSIGHT 2: PRODUCT CATEGORY PERFORMANCE
# ============================================================

category_summary = df.groupby("Product_Category").agg(
    Sales=("Sales_Amount", "sum"),
    Profit=("Profit", "sum"),
    Units_Sold=("Quantity", "sum")
)

category_summary["Profit_Margin_%"] = (
    category_summary["Profit"] /
    category_summary["Sales"] * 100
)

category_summary = category_summary.sort_values(
    "Sales", ascending=False
)

print("\nPRODUCT CATEGORY PERFORMANCE")
print(category_summary.round(2))

best_category = category_summary["Sales"].idxmax()
print("Insight: Best category by sales:", best_category)


# GRAPH 2: SALES BY CATEGORY

category_summary["Sales"].plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()


# ============================================================
# 9. INSIGHT 3: TOP 10 PRODUCTS
# ============================================================

product_summary = df.groupby("Product_Name").agg(
    Sales=("Sales_Amount", "sum"),
    Profit=("Profit", "sum"),
    Units_Sold=("Quantity", "sum")
)

top_products = product_summary.sort_values(
    "Sales", ascending=False
).head(10)

print("\nTOP 10 PRODUCTS BY SALES")
print(top_products.round(2))

print(
    "Insight: Highest-sales product:",
    top_products["Sales"].idxmax()
)


# GRAPH 3: TOP 10 PRODUCTS

top_products["Sales"].sort_values().plot(
    kind="barh",
    figsize=(10, 6)
)

plt.title("Top 10 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product Name")
plt.tight_layout()
plt.show()


# ============================================================
# 10. INSIGHT 4: SALES CHANNEL PERFORMANCE
# ============================================================

channel_summary = df.groupby("Sales_Channel").agg(
    Sales=("Sales_Amount", "sum"),
    Profit=("Profit", "sum"),
    Transactions=("Transaction_ID", "nunique")
)

channel_summary["Profit_Margin_%"] = (
    channel_summary["Profit"] /
    channel_summary["Sales"] * 100
)

channel_summary = channel_summary.sort_values(
    "Sales", ascending=False
)

print("\nSALES CHANNEL PERFORMANCE")
print(channel_summary.round(2))

best_channel = channel_summary["Sales"].idxmax()
print("Insight: Best sales channel:", best_channel)


# GRAPH 4: SALES BY CHANNEL

channel_summary["Sales"].plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Sales by Sales Channel")
plt.xlabel("Sales Channel")
plt.ylabel("Total Sales")
plt.xticks(rotation=25, ha="right")
plt.tight_layout()
plt.show()


# ============================================================
# 11. INSIGHT 5: CUSTOMER SEGMENT ANALYSIS
# ============================================================

segment_summary = df.groupby("Customer_Segment").agg(
    Sales=("Sales_Amount", "sum"),
    Profit=("Profit", "sum"),
    Customers=("Customer_ID", "nunique")
)

segment_summary = segment_summary.sort_values(
    "Sales", ascending=False
)

print("\nCUSTOMER SEGMENT ANALYSIS")
print(segment_summary.round(2))

best_segment = segment_summary["Sales"].idxmax()
print("Insight: Highest-sales customer segment:", best_segment)


# GRAPH 5: SALES BY CUSTOMER SEGMENT

segment_summary["Sales"].plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")
plt.xticks(rotation=25, ha="right")
plt.tight_layout()
plt.show()


# ============================================================
# 12. INSIGHT 6: COUNTRY PERFORMANCE
# ============================================================

country_summary = df.groupby("Country").agg(
    Sales=("Sales_Amount", "sum"),
    Profit=("Profit", "sum")
)

country_summary = country_summary.sort_values(
    "Sales", ascending=False
)

print("\nCOUNTRY PERFORMANCE")
print(country_summary.round(2))

best_country = country_summary["Sales"].idxmax()
print("Insight: Highest-sales country:", best_country)


# GRAPH 6: SALES BY COUNTRY

country_summary["Sales"].plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Sales by Country")
plt.xlabel("Country")
plt.ylabel("Total Sales")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()


# ============================================================
# 13. INSIGHT 7: RETURNS ANALYSIS
# ============================================================

return_flag = df["Return_Flag"].astype("string").str.lower()

returned_rows = df[return_flag == "yes"]
return_rate = (
    len(returned_rows) / len(df) * 100
    if len(df) else np.nan
)

print("\nRETURN ANALYSIS")
print("Returned transaction rows:", len(returned_rows))
print("Return rate (%):", round(return_rate, 2))

print("\nReturn reasons:")
print(returned_rows["Return_Reason"].value_counts())

print("\nReturns by product category:")
print(returned_rows["Product_Category"].value_counts())


# GRAPH 7: RETURN REASONS

returned_rows["Return_Reason"].value_counts().plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Reasons for Product Returns")
plt.xlabel("Return Reason")
plt.ylabel("Number of Transaction Rows")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()


# ============================================================
# 14. INSIGHT 8: MONTHLY SALES TREND
# ============================================================

monthly_sales = (
    df.dropna(subset=["Order_Date"])
      .set_index("Order_Date")["Sales_Amount"]
      .resample("MS")
      .sum()
)

print("\nHIGHEST SALES MONTHS")
print(monthly_sales.sort_values(ascending=False).head(5))

# GRAPH 8: MONTHLY SALES TREND

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# ============================================================
# 15. INSIGHT 9: DISCOUNT AND PROFIT ANALYSIS
# ============================================================

discount_summary = df.groupby(
    "Discount_Percentage"
).agg(
    Average_Profit=("Profit", "mean"),
    Total_Sales=("Sales_Amount", "sum"),
    Transactions=("Transaction_ID", "nunique")
)

print("\nDISCOUNT AND PROFIT ANALYSIS")
print(discount_summary.round(2))

correlation = df[
    ["Discount_Percentage", "Sales_Amount", "Profit"]
].corr()

print("\nCORRELATION MATRIX")
print(correlation.round(3))


# GRAPH 9: DISCOUNT VS AVERAGE PROFIT

discount_summary["Average_Profit"].plot(
    kind="line",
    marker="o",
    figsize=(9, 5)
)

plt.title("Discount Percentage vs Average Profit")
plt.xlabel("Discount Percentage")
plt.ylabel("Average Profit")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()


# ============================================================
# 16. INSIGHT 10: CUSTOMER RATINGS AND DELIVERY
# ============================================================

print("\nCUSTOMER EXPERIENCE")

print(
    "Average customer rating:",
    round(df["Customer_Rating"].mean(), 2)
)

print(
    "Average delivery days:",
    round(df["Delivery_Days"].mean(), 2)
)

delivery_summary = df.groupby("Shipping_Method").agg(
    Average_Delivery_Days=("Delivery_Days", "mean"),
    Average_Rating=("Customer_Rating", "mean")
)

print("\nDELIVERY AND RATING BY SHIPPING METHOD")
print(delivery_summary.round(2))


# GRAPH 10: AVERAGE DELIVERY DAYS

delivery_summary["Average_Delivery_Days"].plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Average Delivery Days by Shipping Method")
plt.xlabel("Shipping Method")
plt.ylabel("Average Delivery Days")
plt.xticks(rotation=25, ha="right")
plt.tight_layout()
plt.show()


# ============================================================
# 17. FINAL AUTOMATIC INSIGHTS
# ============================================================

print("\n" + "=" * 55)
print("FINAL BUSINESS INSIGHTS")
print("=" * 55)

print(f"1. Total sales: {total_sales:,.2f}")
print(f"2. Total profit: {total_profit:,.2f}")
print(f"3. Overall profit margin: {profit_margin:.2f}%")
print(f"4. Total unique orders: {total_orders:,}")
print(f"5. Total unique customers: {total_customers:,}")
print(f"6. Highest-sales year: {best_year}")
print(f"7. Best product category: {best_category}")
print(f"8. Best sales channel: {best_channel}")
print(f"9. Best customer segment: {best_segment}")
print(f"10. Best country by sales: {best_country}")
print(f"11. Return rate by transaction row: {return_rate:.2f}%")

print("\nPROJECT ANALYSIS COMPLETED")
```
````
